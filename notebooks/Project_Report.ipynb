# 📘 Introduction

Molecular symmetry plays a central role in the analysis of molecular structure and in the prediction of spectroscopic properties, including infrared and Raman activity. Symmetry operations, associated with corresponding symmetry elements, are transformations that leave a molecule invariant. The fundamental operations include reflection through a plane, rotation about an n-fold axis, inversion, and improper rotation, which combines rotation followed by reflection. These elements form the basis for the classification of molecules into point groups.

However, the determination of molecular point groups remains a non-trivial and error-prone task, particularly for students and non-experts. The procedure requires the systematic identification of symmetry elements in three-dimensional space, followed by the application of a decision scheme that can be difficult to carry out consistently without strong spatial intuition. In addition, existing computational tools are often primarily focused on automated classification and provide limited pedagogical guidance or interactive support for learning the underlying reasoning process.

In this context, the ACEsymmetry package has been developed to address these limitations by facilitating both the analysis and the learning of molecular symmetry in an educational setting. The package provides tools for visualizing molecules in defined conformations and for displaying their point group together with the associated symmetry elements. It also includes an interactive module designed to support the learning of point group determination through a flowchart-based approach.

## 1. Interface

In order to fulfill the educational objectives of the project, an interactive user interface was developed using Streamlit, an open-source framework for creating web applications in Python.

The interface adopts a multi-page architecture that provides access to the different functionalities of the package through a unified environment. This design facilitates navigation and enhances the user experience by clearly separating the available modules. The general organization of the interactive part of ACEsymmetry is illustrated below:

```python
GUI/                       
│ 
├── Interface.py              # Python file which defines the Home page and contains the code for it 
│ 
└── pages/                    # Folder which will contrain all the pages for the interface 
    ├── Point_group.py        # Module for pure molecular symmetry analysis
    └── Learning_Mode.py      # Pedagogic mode
```

To navigate between the different pages of the ACEsymmetry application, two main options are available.
First, users can access the different modules via the sidebar provided by Streamlit, which automatically lists all pages contained in the pages/ directory.

Alternatively, navigation is also implemented through buttons directly integrated into the Home page, allowing a more guided user experience. These buttons redirect the user to the corresponding module using Streamlit’s built-in page switching function:

```python
st.switch_page("pages/Point_Group")
st.switch_page("pages/Learning_Mode.py")
```

The interface was build using several fundamental Streamlit functions, such as:
- 'st.set_page_config()' to define global page settings such as title, layout, and sidebar behaviour across the application.
- 'st.logo()' to display a clickable institutional logo in the interface header.
- 'st.title()' to structure each page and define clear section headings.
- 'st.button' to develop the interactive character
- 'st.switch_page()' to move through the pages
- 'st.columns()' to organise the layout into multiple sections and improve visual structure and usability

### Page 1 (Point_Group)

The first page of the ACEsymmetry application is dedicated to molecular symmetry analysis and point group determination. It provides a complete workflow allowing the user to define a molecule, generate its 3D structure, and compute its symmetry properties in an interactive way.

The interface is structured as a multi-step process implemented through custom navigation functions ('next_page()' and 'previous_page()'), which rely on 'st.session_state' to maintain the user’s progress across reruns. The page is divided into several sequential sub-sections corresponding to molecule definition, conformer selection, file configuration, and symmetry computation.

#### Molecule input and representation

The first step allows the user to define the studied molecule using one of the three input modes: SMILES notation, IUPAC name, or manual drawing. This is implemented using 'st.radio()' combined with conditional input fields ('st.text_input()') and a molecular editor ('streamlit_ketcher'). Conversion between representations is handled using 'PubChemPy' and custom conversion functions from the acesymmetry.'Format_Conversion module'.

The selected molecule is stored in 'st.session_state.molecule_name', ensuring persistence across the different steps of the interface. In the case of IUPAC input, the name is first converted to SMILES using a conversion function before being stored.

For the drawing mode, the molecule is generated using 'streamlit_ketcher' and then validated through PubChemPy by retrieving the corresponding compound and checking its SMILES representation.

#### Conformer generation

The second step enables the user to define the number of conformers used in the structural search. This is implemented using 'st.radio()' and 'st.slider()', allowing either a default value (1000 conformers) or a user-defined number. The selected value is stored in 'st.session_state.number_of_conformer'.

#### File naming configuration

The third step allows optional renaming of output structure files (.xyz and .SDF). This is implemented using 'st.radio()' and 'st.text_input()'. If no custom name is provided, default filenames are used. These values are also stored in st.session_state to ensure persistence across steps. A basic restriction is mentioned to avoid special characters, although no strict validation is enforced in the code.

#### Point group computation and structure generation

The final step performs the full molecular processing pipeline. When triggered, the function 'conv.overall_conversion_from_smiles()' generates 3D molecular structures and writes the corresponding output files.

The resulting .xyz file is then parsed using 'vd.read_xyz_file()' to extract atomic symbols and coordinates.

The molecular geometry is passed to the pointgroup library, which computes the point group of the molecule. The corresponding symmetry elements are then retrieved using 'vd.get_symmetry_set()'.

#### Output display

The computed point group and symmetry elements are displayed using 'st.write()'. In addition, a structured container ('st.container()') is used to display each symmetry element in a readable format.

Basic error handling is implemented to manage invalid molecules or unsupported cases (for example molecules containing metals or failed conversions), although exceptions are currently handled in a generic way. 

In addition, the program returns a static image of the molecule in a specific conformation.

### Page 2 (Learning Mode)

The ACEsymmetry application also includes a dedicated learning mode for molecular symmetry analysis and point group determination. This module is designed as an interactive educational workflow in which the user is guided through a structured decision process based on a theoretical flowchart, in order to determine the symmetry properties and point group of a molecule.

The interface is implemented as a multi-step Streamlit application, where navigation is controlled by custom functions ('next_page()' and 'previous_page()') relying on 'st.session_state' to preserve the application state across reruns. The workflow is composed of four sequential stages: an introductory flowchart, molecular definition, file configuration, and a guided symmetry classification procedure.

#### Flowchart introduction

The first stage provides an overview of the theoretical decision flowchart used for point group determination. This section introduces the logical structure underlying the classification procedure and establishes the pedagogical framework of the learning mode.

The flowchart is displayed using 'st.image()'. Navigation is handled through a forward progression button or a return mechanism implemented via 'st.switch_page()', allowing the user to exit the learning module.

#### Molecular input and representation

The second stage enables the definition of the molecular system through three distinct input modalities: SMILES notation, IUPAC name, or manual structure drawing. This functionality is implemented using 'st.radio()' combined with conditional input fields ('st.text_input()') and an integrated molecular editor ('streamlit_ketcher').

Conversion between molecular representations is performed using PubChemPy in conjunction with custom conversion routines provided by the 'acesymmetry.Format_Conversion' module. When an IUPAC name is provided, it is systematically converted into a SMILES representation prior to further processing.

In the case of manual drawing, the structure is generated using 'streamlit_ketcher' and validated through PubChemPy by retrieving the corresponding molecular record and extracting its SMILES notation.

The resulting molecular representation is stored in 'st.session_state.molecule_name', ensuring persistence throughout the subsequent steps of the workflow.

#### File naming configuration

The third stage allows optional configuration of output file names for the generated molecular structures (.xyz and .SDF). This is implemented using 'st.radio()' and 'st.text_input()'.

In the absence of user-defined names, default file naming conventions are applied. All file name parameters are stored in 'st.session_state' to ensure consistency across the application workflow. Although constraints regarding allowed characters are indicated in the user interface, no strict validation is enforced in the current implementation. 

#### Guided symmetry decision tree

The final stage implements an interactive decision tree reproducing the theoretical flowchart for point group determination. 

The molecular structure is first generated using 'conv.overall_conversion_from_smiles()', with a fixed conformer count of 1000, chosen as a compromise between computational cost and geometrical reliability of the generated structure. The resulting .xyz file is subsequently parsed using 'vd.read_xyz_file()' in order to extract atomic symbols and Cartesian coordinates.

The molecular geometry is then processed using the 'pointgroup' library to compute the corresponding point group. The associated symmetry elements are obtained via 'vd.get_symmetry_set()'.

The user is subsequently guided through a sequence of binary questions implemented using 'st.radio()', reproducing the logical steps of the flowchart. Each question evaluates the presence of specific symmetry elements, including:

- molecular linearity
- inversion centre
- principal rotation axes (Cn)
- secondary C2 axes
- mirror planes (σh, σv, σd)
- improper rotation axes (Sn)

User responses are systematically compared against the computed symmetry properties using dedicated helper functions such as 'check_linearity()', 'check_inversion_centre()', and 'check_rotation_axis()'.

Immediate feedback is provided after each response, indicating whether the selected answer is consistent with the computed symmetry classification.

#### Symmetry analysis utilities

A set of auxiliary functions is used to process and interpret symmetry elements in a structured manner:

- 'normalize_label()': removes prime notation from symmetry labels
- 'parse_symmetry_element()': extracts multiplicity and symmetry type from symmetry elements
- 'check_linearity()': determines whether the molecule is linear based on its point group
- 'check_inversion_centre()': identifies the presence of an inversion centre
- 'check_rotation_axis()': determines the highest-order principal rotation axis
- 'check_C2_multiplicity()': evaluates the number of perpendicular C2 axes
- 'check_horizontal_plane()', 'check_vertical_plane()', 'check_dihedral_plane()': detect mirror planes
- 'check_improper_rotation_axis()': identifies improper rotation axes (Sn)
- 'check_main_axis_multiplicity()': identifies highly symmetric point groups
- 'check_icosahedral_symmetry()': detects icosahedral symmetry (Ih)

These functions operate on the symmetry set generated by 'vd.get_symmetry_set()' and are used throughout the decision process to validate user responses.

#### Output display

Upon completion of the decision tree, the computed point group is displayed using 'st.write()', along with the complete set of symmetry elements associated with the molecule.

A structured container ('st.container(border=True)') is used to present each symmetry element in a clear and organized format.

Basic error handling is implemented to manage invalid molecular inputs or computational failures (e.g. unsupported structures or conversion errors). However, exceptions are currently handled in a generic manner without detailed classification.

## 2. XYZ File Generator

This module is part of the ACEsymmetry backend and is responsible for molecular data processing and structural conversion. It provides functionalities to convert molecular representations (SMILES, common names, and RDKit Mol objects), detect specific chemical features, and generate 3D molecular structures in both .xyz and .SDF formats.

The module relies on RDKit for molecular manipulation and conformer generation, as well as PubChemPy for chemical name resolution. It is designed as a computational backend and is used by the main Streamlit application for structure generation and symmetry analysis.

#### Molecular name and structure conversion

The function 'smiles_from_name()' retrieves the SMILES representation of a molecule from its common (non-systematic) name using the PubChem database via PubChemPy. If the input is not a string or if the molecule is not found, appropriate exceptions are raised.

The function 'mol_from_smiles()' converts a SMILES string into an RDKit 'Mol' object. Hydrogen atoms are explicitly added using 'Chem.AddHs()' in order to ensure correct 3D structure generation. Invalid SMILES strings return 'None'.

The function 'name_from_mol()' performs the inverse operation by extracting a molecule’s IUPAC name from its RDKit representation. This is achieved by converting the molecule into a SMILES string and querying PubChemPy.

#### Metal detection

The function 'contains_metal()' identifies whether a molecule contains metallic elements based on atomic number classification. A predefined set of atomic numbers corresponding to:

- s-block metals
- transition metals
- post-transition metals
- lanthanides
- actinides

is used to detect coordination compounds.

If at least one atom in the molecule belongs to this set, the function returns 'True'. Otherwise, it returns 'False'. This step is used to exclude coordination complexes from the current computational pipeline.

#### Conformer generation and structure optimisation

The function 'conformer_selection()' is responsible for generating 3D conformations and identifying the lowest-energy structure.

For small molecules (fewer than 3 atoms), a single conformer is generated using 'AllChem.EmbedMolecule()'. For larger molecules, multiple conformers are generated using the ETKDGv3 algorithm via 'AllChem.EmbedMultipleConfs()'.

A force field optimisation is then performed using MMFF94 when possible, otherwise UFF is used as fallback. The energy of each conformer is computed, and the lowest-energy conformer is selected as the most stable structure.

The corresponding structure is exported in SDF format using 'Chem.MolToMolBlock()' and written to file.

If conformer generation fails, the function returns 'None'.

#### Global conversion from SMILES

The function 'overall_conversion_from_smiles()' constitutes the main pipeline of the module. It performs the full conversion from a SMILES string to 3D molecular files.

The procedure is structured as follows:

1. Conversion of SMILES into an RDKit 'Mol' object using 'mol_from_smiles()'
2. Retrieval of the molecule’s name using 'name_from_mol()'
3. Detection of metallic atoms using 'contains_metal()'
4. Generation of the most stable conformer using 'conformer_selection()'
5. Export of the final structure into:
  -.SDF format (via 'conformer_selection()')
  -.xyz format (via 'Chem.MolToXYZBlock()')

If any step fails (invalid SMILES, inability to generate conformers, or presence of unsupported elements), the function raises an exception with a descriptive error message.

The function returns the filenames of the generated .SDF and .xyz files upon successful execution.

#### Conversion from molecular name

The function 'overall_conversion_from_name()' extends the previous pipeline by accepting a common molecule name instead of a SMILES string.

The name is first converted into SMILES using 'smiles_from_name()', after which the standard conversion pipeline 'overall_conversion_from_smiles()' is executed.

This function provides a user-friendly entry point for molecules identified by common names.

#### Module role in the ACEsymmetry application

This module serves as a core computational backend for the ACEsymmetry application. It is responsible for:

- translating chemical identifiers into computational representations
- generating physically relevant 3D structures
- ensuring structural validity before symmetry analysis

It operates independently from the Streamlit interface and provides the necessary molecular data used in downstream symmetry computation and visualization.

## 3. Visualization of the Molecule

This module is part of the ACEsymmetry backend and provides functionalities for symmetry analysis, molecular geometry processing, and visualization of molecular structures. It operates on .xyz files generated by the conversion pipeline and extracts both structural and symmetry-related information for further processing.

The module integrates the pointgroup library for symmetry classification, NumPy for numerical handling, and xyzrender for molecular visualization. It also relies on a custom dictionary 'acesymmetry.Symmetry_Elements_Dico' to map point groups to their corresponding symmetry elements.

#### Symmetry element retrieval from point group

The function 'get_symmetry_set()' retrieves the set of symmetry elements associated with a given point group.

The input point group is validated to ensure it is a string and corresponds to a known key in 'Symmetry_Elements_Dico'. If the point group is not recognized, a 'ValueError' is raised.

This function provides the fundamental link between computed point groups and their associated symmetry elements used throughout the application.

#### XYZ file parsing

The function 'read_xyz_file()' reads molecular geometry from an .xyz file and separates the data into two structured outputs:

- a list of atomic symbols
 -a NumPy array of Cartesian coordinates

The parsing process ignores metadata lines and assumes that the file is located in the current working directory. The correspondence between atoms and coordinates is preserved via indexing.

This function constitutes the base input layer for all subsequent symmetry and structural analyses.

#### Symmetry label extraction

The function 'get_labels()' determines the set of symmetry element labels associated with a molecule described by an .xyz file.

It performs the following steps:

1. Reads atomic symbols and coordinates using 'read_xyz_file()'
2. Computes the point group using 'pg.PointGroup().get_point_group()'
3. Retrieves the symmetry elements using 'get_symmetry_set()'

The resulting set of labels provides a direct representation of the molecule’s symmetry content.

#### Centre of mass computation

The function 'get_barycentre()' computes the centre of mass of a molecule using atomic symbols and Cartesian coordinates extracted from an .xyz file.

The calculation is performed using 'pg.tools.get_center_mass()'. The resulting coordinates are written to a new .xyz file containing a single atom labeled as 'G', representing the barycentre.

The function returns the computed centre of mass as a list of coordinates.

#### Inversion centre detection

The function 'get_inversion_centre()' determines whether a molecule possesses an inversion centre.

The point group is first computed using 'pg.PointGroup().get_point_group()', and the corresponding symmetry elements are retrieved using 'get_symmetry_set()'.

If the inversion symmetry element ('i') is present, the function returns the barycentre coordinates as an approximation of the inversion centre.

Otherwise, the function indicates that no inversion centre is present and returns 'None'.

This approach assumes equivalence between the molecular centre of mass and inversion centre when symmetry is present.

#### Molecular visualization

The function 'display()' generates a graphical representation of a molecule from its .xyz file.

The molecule is loaded using 'xyzrender.load()' and rendered using 'xyzrender.render()'. The output is saved as an image file, with .png used by default if no filename is provided.

Hydrogen atoms are included in the visualization, and a predefined rendering configuration ('Pmol') is applied.

The function returns the filename of the generated image.

#### Visualization with centre of mass overlay (experimental)

The function 'display_with_mass_centre()' extends the standard visualization by overlaying the molecular centre of mass onto the rendered image.

The barycentre is first computed using 'get_barycentre()', and an additional .xyz file is created to represent it as a pseudo-atom.

The molecule is then rendered using 'xyzrender.render(), including an overlay of the barycentre file.

However, this functionality is currently experimental due to limitations in xyzrender:

-the use of a non-standard atomic symbol (e.g. 'G') may cause rendering errors
-alternative substitutions may lead to structural inconsistencies during rendering

As a result, this feature is not fully robust in the current implementation.

#### Module role in the ACEsymmetry architecture

This module serves as a post-processing and visualization layer within the ACEsymmetry application. It complements the molecular conversion and symmetry computation modules by providing:

- parsing of .xyz molecular structures
- extraction of symmetry elements from computed point groups
- computation of molecular centre of mass
- graphical rendering of molecular structures

It is primarily used for analysis validation and visual interpretation of molecular symmetry properties.

# 🔬 Discussion

## Strengths

### 1. Intermodality

The ACEsymmetry package is designed as a versatile tool that combines both an analysis mode and a learning mode within a single application. This dual structure allows the software to address different user profiles, ranging from students to more experienced users in computational or theoretical chemistry.

In an educational context, particularly in EPFL courses such as *Mathematical Methods in Chemistry*, the application provides an interactive complement to traditional teaching methods. Instead of relying solely on static representations or manual determination of point groups, students can actively engage with molecular structures and immediately visualize the consequences of symmetry operations.

More generally, the learning mode transforms an otherwise abstract and algorithmic procedure into a guided step-by-step reasoning process, which helps reinforce conceptual understanding. At the same time, the analysis mode provides a fully automated computational pipeline, making the tool relevant for both pedagogical and exploratory research purposes. 

### 2. Interactivive character

The ACEsymmetry package is implemented as a Streamlit-based web application, accessible through a browser after installation. This design significantly improves usability compared to classical command-line tools, particularly for users without programming experience.

One of the main advantages of the interface lies in its multimodal input system, which allows users to define molecules through IUPAC names, SMILES notation, or direct graphical drawing. This flexibility reduces the entry barrier and makes the tool accessible to a wider audience with different backgrounds in chemistry and informatics.

Beyond input flexibility, the interactive structure ensures immediate feedback at each step of the workflow. Users can directly observe how changes in molecular representation influence conformer generation, geometry optimization, and ultimately symmetry classification. This real-time feedback loop enhances both usability and educational value.


## Limitations

### 1. Limited chemistry

The conformer generation and geometry optimization in this work are performed using RDKit, relying primarily on the MMFF94 force field, with UFF used as a fallback when MMFF94 parameters are not available. While this approach is efficient and well-suited for general organic molecules, it is based on empirical potential energy functions rather than quantum mechanical calculations.

As a consequence, these force fields mainly capture steric effects and bonded interactions, but do not explicitly account for the electronic structure of molecules. This approximation limits the accuracy of the method in systems where electronic effects play a significant role in determining molecular geometry.

In particular, the approach is not suitable for metallic coordination complexes, which often require a more detailed treatment of electronic interactions and are therefore excluded from the computational workflow. Similarly, molecules exhibiting strong electron delocalization, conjugation, or orbital-driven structural effects may not be accurately represented in their lowest-energy conformations.

Overall, while the method provides a robust and computationally efficient solution for standard organic systems, its applicability remains limited in chemically complex systems where electronic structure effects dominate molecular geometry.

### 2. Limited visualization

A current limitation of the application is its inability to display the molecular structure and its associated symmetry elements within the same graphical representation. While the program is able to compute the point group and identify the corresponding symmetry elements, these elements cannot be directly visualized alongside the molecular geometry.

As a consequence, the symmetry operations remain abstract for the user, since their spatial relationship with the molecule cannot be explicitly illustrated. This prevents a direct visual interpretation of key symmetry features such as rotation axes, mirror planes, or inversion centers in relation to the atomic structure.

From a pedagogical perspective, this limitation reduces the explanatory power of the tool, as users must rely on textual output rather than geometric visualization to understand the symmetry properties of the molecule.

## Further improvements

### 1. Chemistry enlarging

As previously discussed, the conformer generation module is limited by its inability to accurately describe systems where electronic effects play a dominant role, such as metallic coordination complexes or molecules with non-covalent interaction-driven geometries.

A possible improvement would be to extend the applicability of the application to such systems by incorporating more advanced computational chemistry methods. This was not implemented in the present work due to their increased complexity and significantly higher computational cost.

In particular, a more accurate description of molecular geometry can be achieved through quantum chemistry approaches, such as Density Functional Theory (DFT), which solve the electronic Schrödinger equation numerically. While these methods provide a more reliable representation of electronic effects, they require substantially greater computational resources and more complex software integration.

Integrating such approaches into ACEsymmetry would therefore represent a significant but promising extension of the current pipeline.

### 2. Complete and interactive visualization

The visual output of the application represents an important area for improvement. As mentioned previously, one possible extension would be the direct representation of symmetry elements within the molecular structure. This would significantly enhance the interpretability of the computed point group by making abstract symmetry operations directly visible.

In addition, the visualization could be enriched by highlighting symmetry-equivalent atoms. Such a feature would improve the chemical interpretability of the results and could provide useful insights into spectroscopic properties, particularly in relation to NMR spectroscopy. Indeed, the identification of equivalent атом sets is directly related to the number of signals observed in ¹³C and ¹H NMR spectra.

Furthermore, the integration of an interactive 3D molecular viewer, allowing real-time rotation and manipulation of the structure, would improve the understanding of spatial symmetry. This would enable users to better identify symmetry elements and observe how deviations from ideal symmetry affect the molecular structure.
